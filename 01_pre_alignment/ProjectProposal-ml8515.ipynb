### Identification of IL13-Mediated Chromatin and Transcriptional Reprogramming of Epithelial Gene Expression in Eosinophilic Esophagitis

Megha Lal  
Jan 22, 2026

#### Problem Statement 
Eosinophilic esophagitis (EoE) is a chronic, non-IgE-mediated food allergy characterized by eosinophilic inflammation of the esophagus and profound epithelial barrier dysfunction driven by type 2 responses. Despite treatment reducing immune cell infiltration, epithelial abnormalities may persist even in histologic remission. Recent transcriptomic studies indicate that patients with post-treatment EoE epithelium continue to display altered gene expression compared to normal esophageal tissue, even at strict histological cutoffs for deep remission (eos/hpf = 0). This suggests that inflammatory exposure may induce durable regulatory changes beyond transient immune activation.

Interleukin-13 (IL-13) is a central effector cytokine in EoE pathogenesis and is sufficient to recapitulate hallmark epithelial features of the disease in vitro, including barrier dysfunction and disease-associated transcriptional signatures. IL-13 exposure induces chromatin-level alterations, such as increased histone acetylation at loci including CCL26, and drives STAT6-dependent upregulation of epithelial genes such as CAPN14 and NTRK1. Despite these findings, critical gaps remain in our understanding of IL-13–driven epithelial reprogramming. Specifically,
* how IL-13 reshapes chromatin accessibility landscape genome-wide.
* whether these chromatin changes persist after cytokine withdrawal.
* how distinct inflammatory exposure histories give rise to stable versus reversible epithelial states.


#### Biological Significance
Defining how IL-13 reprograms epithelial regulatory landscapes is essential to understanding the persistence of epithelial dysfunction in EoE. While RNA-seq captures downstream transcriptional consequences of inflammation, it cannot distinguish transient cytokine responses from stable regulatory states encoded at the level of chromatin accessibility. IL-13–induced changes in chromatin structure may establish a form of epithelial “epigenetic memory,” predisposing the tissue to persistent dysfunction even after inflammatory signals are withdrawn. By integrating ATAC-seq and RNA-seq in physiologically relevant epithelial models, this study will delineate reversible versus persistent IL-13–driven regulatory programs, identify transcription factors that mediate these effects, and provide mechanistic insight into why epithelial abnormalities may endure despite histologic remission. These findings have the potential to inform therapeutic strategies aimed at restoring durable epithelial homeostasis in EoE.

#### Data and Resources
We hypothesize that IL-13 induces stable changes in chromatin accessibility that drive sustained transcriptional reprogramming of epithelial barrier genes. These accessibility states may persist after cytokine withdrawal and contribute to long-term epithelial dysfunction in EoE. This mechanism could explain why patients with EoE often exhibit ongoing epithelial dysfunction despite achieving histologic remission.

**Experimental Setup**
To test this hypothesis, we will leverage telomerase-immortalized human esophageal keratinocytes (EPC2-hTERT) cultured as both 2D monolayer cultures and 3D esophageal organoids. Organoids provide a physiologically relevant model that recapitulates growth, differentiation, and barrier function of the human esophageal epithelium.

* First-Generation Organoids: Acute IL-13 effects on chromatin and gene expression
    * **NS (baseline)**: Organoids differentiated in normal media.
    * **IL-13 (acute exposure)**: Organoids exposed to IL-13 during differentiation to model acute cytokine exposure.

 * Second-Generation Organoids: Persistence and reversibility of IL-13–induced programs
    * **NS-NS (baseline)**: Monolayers maintained in normal media, transitioned into 3D organoids without IL-13 exposure.
    * **IL13-NS (memory after withdrawal)**: Monolayers treated with IL-13, transitioned into 3D organoids, and cultured without IL-13 to assess persistence or reversibility of regulatory changes.
    * **IL13-IL13 (chronic exposure)**: Monolayers treated with IL-13, transitioned into 3D organoids, and continuously exposed to IL-13 to mimic sustained type 2 inflammation.

**Data Resources**  
Bulk RNA-seq and bulk ATAC-seq data have been generated in-house from EPC2-hTERT–derived 3D organoids across acute, chronic, and withdrawal IL-13 conditions. 

This experimental framework and paired datasets enables direct comparison of acute induction, chronic reinforcement, and post-withdrawal persistence of IL-13–mediated chromatin and transcriptional states.

#### Analytical Tools and Methods 

ATAC-seq Workflow:
* **Preprocessing**: Quality assessment via *FastQC* and adapter trimming with *Trim Galore*.
* **Alignment & Filtering**: Alignment to GRCh38.p13 using *Bowtie2*. PCR duplicates, mitochondrial reads (chrM), and ENCODE blacklist regions will be removed to refine the signal using *Picard* and *Samtools*.
* **Quality Metrics**: Library integrity will be confirmed through fragment length periodicity, TSS enrichment scores, and FRiP (Fraction of Reads in Peaks).
* **Peak Calling**: Accessible regions will be identified using *MACS2* (q ≤ 0.05). A consensus peak set will be merged across all samples to enable uniform quantification.
* **Differential Accessibility**: Comparative analysis between NS-NS, IL13-IL13, and IL13-NS groups will be performed using *DiffBind* (FDR ≤ 0.05).
* **Motif Analysis**: Condition-specific transcription factor activity will be inferred using *chromVAR* and the *JASPAR2020* motif database.

RNA-seq Workflow:
* **QC & Trimming**: Initial quality assessment via *FastQC* and contaminant screening with *FastQScreen*. Adapters and low-quality bases will be removed using *Trimmomatic*.
* **Quantification**: Transcript abundance will be estimated using *kallisto* against the GRCh38.p13 reference transcriptome.
* **Preprocessing**: Low-abundance genes will be filtered to increase statistical power. Reads will be normalized using the TMM (Trimmed Mean of M-values) method via *edgeR*.
* **Differential Expression**: Dimensionality reduction via *PCA* will assess sample clustering. Differential expression analysis across conditions will be performed using *limma-voom*.
* **Functional Analysis**: Pathway enrichment and biological signatures will be identified using *GSEA* (clusterProfiler + MSigDB).

Integration of ATAC-seq & RNA-seq:
* **Peak Annotation**: Differentially accessible regions (DARs) will be annotated using *ChIPseeker*, categorizing peaks as Promoters (<3 kb from TSS) or Distal Regulatory Regions (>3 kb from TSS).
* **Regulatory Association**: Putative gene-peak pairs will be inferred by integrating DARs and DEGs within a 100 kb window, prioritizing links with concordant directionality (e.g., increased accessibility coupled with increased expression).
* **Correlation Analysis**: Coordinated regulatory shifts will be validated by performing Pearson/Spearman correlation between promoter/enhancer accessibility and target gene expression across all experimental conditions.TF.
* **Inference**: Candidate master regulators, such as STAT6, will be identified by cross-referencing motif enrichment in DARs with the expression levels of their corresponding transcription factors.

#### Rationale for Analysis
Characterizing IL-13–mediated transcriptional and chromatin reprogramming is essential to understanding the persistence of epithelial dysfunction in EoE. By integrating RNA-seq and ATAC-seq data across organoid models and primary patient-derived cells, this study will:

* Infer transcription factor activity through integrated motif accessibility and expression changes, highlighting master regulators of epithelial dysfunction.
* Differentiate between reversible and persistent inflammatory programs, providing insight into whether IL-13–induced epigenetic modifications can be reset after cytokine removal or if they represent stable reprogramming 
* Advance therapeutic strategies by identifying candidate pathways and TF networks that could be targeted to restore barrier function in EoE.
